```python
def pop_trailing_group(text: str):
    """Pop the last top-level balanced (...) group off the end of text.
    Returns (group_content, remainder) or (None, text) if text doesn't
    end in a balanced parenthetical.
    """
    text = text.rstrip()
    if not text.endswith(')'):
        return None, text
    close_idx = len(text) - 1
    balance = 0
    open_idx = None
    for i in range(close_idx, -1, -1):
        ch = text[i]
        if ch == ')':
            balance += 1
        elif ch == '(':
            balance -= 1
            if balance == 0:
                open_idx = i
                break
    if open_idx is None:
        return None, text
    group = text[open_idx + 1:close_idx].strip()
    remainder = text[:open_idx].strip()
    return group, remainder


def clean(val):
    """Treat missing/placeholder values ('-', '', None) as None."""
    if val is None:
        return None
    val = val.strip()
    return val if val and val != '-' else None


def split_name_agency_code(text: str):
    """Split 'Project Name (Agency) (Project Code) (Legacy OCMS Code) (PMGID)'.

    The last two fields (Legacy OCMS Code, PMGID) are optional and are
    frequently absent entirely or present as a placeholder '-'. Agency
    names may contain nested parentheses (e.g. '(BCCL)'), and project
    names occasionally end in their own parenthetical (e.g. '(3x800 MW)')
    — both are handled correctly by only ever popping up to 4 trailing
    top-level groups (PMGID, OCMS Code, Project Code, Agency) off the
    end of the string.
    """
    if pd.isna(text):
        return pd.Series([None, None, None, None, None])

    text = text.replace('\n', ' ').strip()

    # Pop up to 4 trailing top-level parenthetical groups from the right.
    groups = []
    remainder = text
    for _ in range(4):
        g, remainder = pop_trailing_group(remainder)
        if g is None:
            break
        groups.append(g)

    n = len(groups)
    if n < 2:
        # Malformed / unexpected format — can't reliably split.
        return pd.Series([text if text else None, None, None, None, None])

    # groups[0] is the rightmost popped group, groups[-1] the leftmost.
    agency = groups[n - 1]
    project_code = groups[n - 2]
    ocms_code = groups[n - 3] if n >= 3 else None
    pmgid = groups[n - 4] if n >= 4 else None

    project_name = remainder.strip() or None

    return pd.Series([
        project_name,
        clean(agency),
        clean(project_code),
        clean(ocms_code),
        clean(pmgid),
    ])
```

```python
df[['project_name', 'agency', 'project_code', 'ocms_code', 'pmgid']] = (
    df['Project Name (Agency) (Project Code)'].apply(split_name_agency_code)
)
df.head()
```

```python
df.drop("Project Name (Agency) (Project Code)", axis =1, inplace=True)
```

```python
parts = df['Date of Approval (Start Date) MM/YYYY'].str.replace(')', '', regex=False).str.split('(', expand=True)
df['date_of_approval'] = parts[0].str.strip().replace('', np.nan)
df['start_date'] = parts[1].str.strip()
df.head()
```

```python
df.drop("Date of Approval (Start Date) MM/YYYY", axis =1, inplace=True)
df.head()
```

```python
parts = (
    df['Orignal/Target DoC (Revised DoC) MM/YYYY']
    .str.replace(')', '', regex=False)
    .str.split('(', expand=True)
)
df['target_doc'] = parts[0].str.strip().replace('', np.nan)
df['revised_doc'] = parts[1].str.strip().replace('', np.nan)
df.head()
```

```python
df.drop("Orignal/Target DoC (Revised DoC) MM/YYYY", axis =1, inplace=True)
df.head()
```

```python
parts = df['Orignal Cost (Revised Cost) in Rs. Crore'].str.replace(')', '', regex=False).str.split('(', expand=True)
df['original_cost_cr'] = parts[0].str.strip()
df['revised_cost_cr'] = parts[1].str.strip()
df.drop("Orignal Cost (Revised Cost) in Rs. Crore", axis =1, inplace=True)
df.head()
```

```python
date_cols = ['date_of_approval', 'start_date', 'target_doc', 'revised_doc']
missing_tokens = {'-', 'NA', 'na', 'N/A', ''}

for c in date_cols:
    df[c] = df[c].replace(missing_tokens, np.nan)
    df[c] = pd.to_datetime(df[c], format='%m/%Y', errors='coerce')
```

```python
cost_cols = ['original_cost_cr', 'revised_cost_cr']
for c in cost_cols:
    df[c] = df[c].replace(missing_tokens, np.nan).astype(float)
```
