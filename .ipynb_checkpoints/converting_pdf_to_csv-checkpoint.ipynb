```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pdfplumber
import os
import glob
```

```python
def clean_table(table):
    """Clean newlines and None values from extracted table cells."""
    cleaned = []
    for row in table:
        cleaned_row = [str(cell).replace('\n', ' ').strip() if cell is not None else "" for cell in row]
        if any(cleaned_row):
            cleaned.append(cleaned_row)
    return cleaned

def extract_project_data(pdf_paths):
    completed_rows = []
    ongoing_rows = []
    
    completed_header = None
    ongoing_header = None

    for pdf_path in pdf_paths:
        print(f"Processing: {pdf_path}")
        try:
            with pdfplumber.open(pdf_path) as pdf:
                current_ministry = ""
                current_sector = ""
                
                for page in pdf.pages:
                    text = page.extract_text()
                    if not text:
                        continue
                    
                    is_completed = "Completed Projects" in text and "Sl.No" in text
                    is_ongoing = "All Ongoing Projects" in text and "Sl.No" in text
                    
                    if is_completed or is_ongoing:
                        tables = page.extract_tables()
                        for table in tables:
                            cleaned_table = clean_table(table)
                            
                            for row in cleaned_table:
                                col0 = row[0].lower().replace(".", "").replace(" ", "")
                                
                                # Capture the table header
                                if "slno" in col0:
                                    # Always keep the longest header found to avoid malformed truncated headers
                                    if is_completed:
                                        if completed_header is None or len(row) > len(completed_header) - 2:
                                            completed_header = row + ["Ministry", "Sector"]
                                    elif is_ongoing:
                                        if ongoing_header is None or len(row) > len(ongoing_header) - 2:
                                            ongoing_header = row + ["Ministry", "Sector"]
                                    continue
                                
                                # DETECT RED ROWS (Ministry & Sector headers)
                                col1 = row[1] if len(row) > 1 else ""
                                other_cols_empty = all(c == "" for c in row[2:]) if len(row) > 2 else True
                                
                                if row[0] == "" and col1 != "" and other_cols_empty:
                                    if "ministry" in col1.lower() or "department" in col1.lower():
                                        current_ministry = col1
                                        current_sector = "" # Reset the sector when a new Ministry begins
                                    else:
                                        current_sector = col1
                                            
                                # DETECT DATA ROWS (Actual Project Entries)
                                elif row[0] != "" and row[0].isdigit():
                                    row.extend([current_ministry, current_sector])
                                    
                                    if is_completed:
                                        completed_rows.append(row)
                                    elif is_ongoing:
                                        ongoing_rows.append(row)
                                        
        except Exception as e:
            print(f"Error processing {pdf_path}: {e}")

    def save_dataframe(rows, header, filename):
        if not rows:
            print(f"No data found for {filename}.")
            return
            
        df = pd.DataFrame(rows)
        # The number of base columns is total columns minus the 2 we manually appended
        base_col_count = len(df.columns) - 2
        
        if header and len(header) >= 2:
            base_header = header[:-2]
        else:
            base_header = []
            
        # Dynamically pad or truncate the header to perfectly match the dataframe size
        if len(base_header) < base_col_count:
            base_header += [f"Unnamed_{i}" for i in range(len(base_header) + 1, base_col_count + 1)]
        else:
            base_header = base_header[:base_col_count]
            
        df.columns = base_header + ["Ministry", "Sector"]
        df.to_csv(filename, index=False)
        print(f"✅ Successfully extracted {len(df)} rows to {filename}")

    # Process and Export
    save_dataframe(completed_rows, completed_header, "data/completed_projects.csv")
    save_dataframe(ongoing_rows, ongoing_header, "data/ongoing_projects.csv")

if __name__ == "__main__":
    pdf_files = glob.glob("dataset/*.pdf")
    if not pdf_files:
        print("No PDF files found. Please place this script in the folder containing your PDFs.")
    else:
        print(f"Found {len(pdf_files)} PDFs. Starting extraction...")
        extract_project_data(pdf_files)
```
