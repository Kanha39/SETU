## For Monthly dataset pdfs

```python
def clean_table(table):
    """Clean newlines and None values from extracted table cells."""
    cleaned = []
    for row in table:
        cleaned_row = [str(cell).replace('\n', ' ').strip() if cell is not None else "" for cell in row]
        if any(cleaned_row):
            cleaned.append(cleaned_row)
    return cleaned

def extract_project_data(pdf_paths):
    completed_rows = []
    ongoing_rows = []
    
    completed_header = None
    ongoing_header = None

    for pdf_path in pdf_paths:
        print(f"Processing: {pdf_path}")
        try:
            with pdfplumber.open(pdf_path) as pdf:
                current_ministry = ""
                current_sector = ""
                
                for page in pdf.pages:
                    text = page.extract_text()
                    if not text:
                        continue
                    
                    is_completed = "Completed Projects" in text and "Sl.No" in text
                    is_ongoing = "All Ongoing Projects" in text and "Sl.No" in text
                    
                    if is_completed or is_ongoing:
                        tables = page.extract_tables()
                        for table in tables:
                            cleaned_table = clean_table(table)
                            
                            for row in cleaned_table:
                                col0 = row[0].lower().replace(".", "").replace(" ", "")
                                
                                # Capture the table header
                                if "slno" in col0:
                                    # Always keep the longest header found to avoid malformed truncated headers
                                    if is_completed:
                                        if completed_header is None or len(row) > len(completed_header) - 2:
                                            completed_header = row + ["Ministry", "Sector"]
                                    elif is_ongoing:
                                        if ongoing_header is None or len(row) > len(ongoing_header) - 2:
                                            ongoing_header = row + ["Ministry", "Sector"]
                                    continue
                                
                                # DETECT RED ROWS (Ministry & Sector headers)
                                col1 = row[1] if len(row) > 1 else ""
                                other_cols_empty = all(c == "" for c in row[2:]) if len(row) > 2 else True
                                
                                if row[0] == "" and col1 != "" and other_cols_empty:
                                    if "ministry" in col1.lower() or "department" in col1.lower():
                                        current_ministry = col1
                                        current_sector = "" # Reset the sector when a new Ministry begins
                                    else:
                                        current_sector = col1
                                            
                                # DETECT DATA ROWS (Actual Project Entries)
                                elif row[0] != "" and row[0].isdigit():
                                    row.extend([current_ministry, current_sector])
                                    
                                    if is_completed:
                                        completed_rows.append(row)
                                    elif is_ongoing:
                                        ongoing_rows.append(row)
                                        
        except Exception as e:
            print(f"Error processing {pdf_path}: {e}")

    def save_dataframe(rows, header, filename):
        if not rows:
            print(f"No data found for {filename}.")
            return
            
        df = pd.DataFrame(rows)
        # The number of base columns is total columns minus the 2 we manually appended
        base_col_count = len(df.columns) - 2
        
        if header and len(header) >= 2:
            base_header = header[:-2]
        else:
            base_header = []
            
        # Dynamically pad or truncate the header to perfectly match the dataframe size
        if len(base_header) < base_col_count:
            base_header += [f"Unnamed_{i}" for i in range(len(base_header) + 1, base_col_count + 1)]
        else:
            base_header = base_header[:base_col_count]
            
        df.columns = base_header + ["Ministry", "Sector"]
        df.to_csv(filename, index=False)
        print(f"✅ Successfully extracted {len(df)} rows to {filename}")

    # Process and Export
    save_dataframe(completed_rows, completed_header, "data/raw/monthly_completed_projects.csv")
    save_dataframe(ongoing_rows, ongoing_header, "data/raw/monthly_ongoing_projects.csv")

pdf_files = glob.glob("dataset/monthly/*.pdf")
if not pdf_files:
    print("No PDF files found. Please place this script in the folder containing your PDFs.")
else:
    print(f"Found {len(pdf_files)} PDFs. Starting extraction...")
    extract_project_data(pdf_files)
```

```python
GROUPING_COLUMN_NAMES = {"state", "state/ut", "state/uts", "sector", "sector name"}
 
 
# --------------------------------------------------------------------------
# 1 & 2. Recover the State column at extraction time
# --------------------------------------------------------------------------
 
def get_state_column_anchors(page, state_x_max, header_y_max=100, line_gap=15):
    """
    Return [(top, state_name), ...] for every state label printed in the
    left-hand State column of this page, in top-to-bottom order.
 
    Each label is drawn once, at the top of the block of rows it applies
    to (a block can span the rest of the page and continue for several
    more pages before the next label appears). A label can wrap onto two
    lines (e.g. "ARUNACHAL" / "PRADESH"); adjacent words within
    `line_gap` points of each other vertically are treated as the same
    label, while a bigger vertical gap marks the start of a new one.
 
    state_x_max is the right edge of the State column (read off the
    header row's cell boundaries: ~71.6 for Table:-6, ~78 for Table:-7 -
    both tables put the header word "State" starting at x0≈14.8).
    header_y_max excludes the repeated column-header text ("State") that
    prints near the top of every page.
    """
    words = [w for w in page.extract_words() if w["x0"] < state_x_max and w["top"] > header_y_max]
    words.sort(key=lambda w: w["top"])
 
    anchors = []
    cur_words, prev_bottom = [], None
    for w in words:
        if prev_bottom is not None and w["top"] - prev_bottom > line_gap:
            anchors.append((cur_words[0]["top"], " ".join(x["text"] for x in cur_words)))
            cur_words = []
        cur_words.append(w)
        prev_bottom = w["bottom"]
    if cur_words:
        anchors.append((cur_words[0]["top"], " ".join(x["text"] for x in cur_words)))
    return anchors
 
 
def inject_state_into_rows(page, rows, state_col_idx=0, header_rows=1, state_x_max=None):
    """
    Overlay real state names onto `rows` (the list of row-lists already
    produced by page.extract_tables() for this page), in place.
 
    rows[header_rows:] must be in the same top-to-bottom order as
    page.find_tables()[0].rows (true for a plain page.extract_tables()
    call, since both come from the same underlying Table object) - that
    lets each anchor's y-position be matched to the data row it belongs
    to, the same row a human reading the page would associate it with.
 
    Only rows whose current state_col_idx value is empty are overwritten,
    so this never clobbers a value that was somehow already captured.
    """
    tables = page.find_tables()
    if not tables:
        return rows
    table = tables[0]
 
    if state_x_max is None:
        # Read the State column's right edge straight off this page's own
        # header cell instead of hard-coding it, so the function keeps
        # working if a future report shifts the table layout slightly.
        header_cells = table.rows[0].cells
        state_x_max = header_cells[state_col_idx][2] if header_cells and header_cells[state_col_idx] else 78
 
    anchors = get_state_column_anchors(page, state_x_max)
    if not anchors:
        return rows
 
    data_rows = table.rows[header_rows:]
    for anchor_top, label in anchors:
        # Find the first data row whose vertical span contains (or starts
        # just after) this anchor - i.e. the row the label sits next to.
        target_idx = None
        for i, r in enumerate(data_rows):
            if r.bbox[3] > anchor_top:  # row's bottom edge is past the anchor
                target_idx = i
                break
        if target_idx is None:
            target_idx = len(data_rows) - 1
 
        row_idx = header_rows + target_idx
        if row_idx < len(rows) and not str(rows[row_idx][state_col_idx] or "").strip():
            rows[row_idx][state_col_idx] = label
 
    return rows
 
 
# --------------------------------------------------------------------------
# 3. process_and_save - same logic, commented for overall functionality
# --------------------------------------------------------------------------
 
def process_and_save(rows, filename):
    """
    Turn the raw per-page table rows collected for one report table (e.g.
    Completed, Frozen, or Ongoing projects) into a single clean DataFrame
    and write it to CSV.
 
    Expects `rows` to already have any position-only data (like the State
    column above) merged in - this function only works with what's in the
    row cells, it can't recover text pdfplumber's grid never captured.
    """
 
    if not rows:
        print(f"No data for {filename}")
        return None
 
    # Pages can yield rows with different lengths (e.g. a continuation
    # page dropping a repeated header cell, or a merged cell shifting a
    # column). Pad every row to the widest row's length before building
    # the DataFrame so pandas doesn't silently shift columns out of
    # alignment when row lengths disagree.
    max_len = max(len(r) for r in rows)
    rows = [list(r) + [""] * (max_len - len(r)) for r in rows]
 
    df = pd.DataFrame(rows)
 
    # STEP 1: Locate the header row. Tables in this report repeat their
    # header on every page, so the "real" header could be anywhere in the
    # first few rows depending on how pages were concatenated. "Sl No" /
    # "Sl.No" / "SI No" (an OCR/text-extraction variant) reliably marks
    # it; check the first 4 columns since some tables (e.g. Ongoing) have
    # State/Sector columns pushed further left than Sl No.
    header_row_idx = None
    for idx, row in df.iterrows():
        cells = [str(c).lower() for c in row[:4]]
        if any("sl" in c or "si no" in c for c in cells):
            header_row_idx = idx
            break
 
    if header_row_idx is not None:
        header_values = [str(c).strip() for c in df.iloc[header_row_idx].tolist()]
        df.columns = header_values
    else:
        header_values = [str(c).strip() for c in df.iloc[0].tolist()]
        df.columns = header_values
        df = df[1:]
 
    # STEP 2: Every page repeats the header, so drop each full-row
    # duplicate of it (matched against the whole row, not just column 0 -
    # column 0 may legitimately repeat real values, e.g. the same state
    # name on consecutive rows, so a column-0-only match would wrongly
    # drop real data).
    header_tuple = tuple(header_values)
    is_header_row = df.apply(lambda r: tuple(str(c).strip() for c in r) == header_tuple, axis=1)
    df = df[~is_header_row]
 
    # A page break can also split the header across two rows, leaving a
    # stray fragment (just "Sl No" etc.) in column 0 on its own row -
    # catch those separately since they won't match the full-row check.
    df = df[~df.iloc[:, 0].astype(str).str.strip().str.lower().isin(["sl no", "sl.no", "sl.no.", "si no"])]
 
    df = df.reset_index(drop=True)
 
    # STEP 3: Forward-fill every grouping column present (State, Sector,
    # ...). These columns hold a value only on the first row of each
    # group and are blank/merged-looking for the rest of that group's
    # rows, so ffill() reconstructs the value for every row. This is only
    # correct if the FIRST row of each group actually has the value
    # captured in `rows` - if page.extract_tables() (or an upstream
    # merged-cell layout) never captures it for any row, ffill has
    # nothing to propagate and the column comes out empty. See
    # inject_state_into_rows() above for recovering such columns before
    # they reach this function.
    grouping_cols = [c for c in df.columns if str(c).lower().strip() in GROUPING_COLUMN_NAMES]
 
    for col in grouping_cols:
        df[col] = df[col].replace('', np.nan)
        df[col] = df[col].infer_objects()
 
        non_null_before = df[col].notna().sum()
        if non_null_before == 0:
            # Nothing to fill from - ffill would silently return an
            # all-empty column. Surface this loudly instead of writing a
            # broken CSV, since it means the extraction step upstream
            # failed to capture this column for every single row.
            print(f"WARNING: '{col}' has no captured values in {filename} - "
                  f"ffill skipped, column will stay empty. Check the "
                  f"upstream extraction for this column.")
            continue
 
        df[col] = df[col].ffill()
 
    if grouping_cols:
        print(f"Filled grouping column(s): {grouping_cols}")
    else:
        print("Could not find any grouping column (State/Sector) to fill")
 
    # STEP 4: Drop rows that are ONLY a group label - i.e. every column
    # except the grouping columns is blank. This covers both standalone
    # sector-title rows (Completed/Frozen) and standalone state/sector
    # title rows on the Ongoing tables.
    other_cols = [c for c in df.columns if c not in grouping_cols]
    if other_cols:
        is_blank_elsewhere = (
            df[other_cols]
            .apply(lambda s: s.astype(str).str.strip())
            .eq('')
            .all(axis=1)
        )
        df = df[~is_blank_elsewhere]
 
    df = df.reset_index(drop=True)
 
    # STEP 5: Normalize column names (multi-line header cells come
    # through with embedded newlines).
    df.columns = [str(col).replace('\n', ' ').replace('\r', ' ').strip() for col in df.columns]
 
    # Save
    output_path = f"data/raw/{filename}"
    df.to_csv(output_path, index=False)
 
    print(f"{filename}")
    print(f"Rows: {len(df)}")
    print(f"Columns: {list(df.columns[:4])}...")
 
    return df
```

```python
if __name__ == "__main__":
    import pdfplumber
 
    PDF_PATH = "dataset/quaterly/QPISR_QR_1st_2025-26.pdf"
    # (start_page, end_page_exclusive, output_filename)
    ONGOING_TABLES = [
        (11, 25, "quaterly_completed_projects.csv"),
        (31, 33, "quaterly_frozen_or_deleted_projects.csv"),
        (58, 256, "quaterly_ongoing_projects.csv"),
    ]
 
    import os
    os.makedirs("data/raw", exist_ok=True)
 
    with pdfplumber.open(PDF_PATH) as pdf:
        for start, end, filename in ONGOING_TABLES:
            all_rows = []
            for page_num in range(start, end):
                page = pdf.pages[page_num]
                page_tables = page.extract_tables()
                if not page_tables:
                    continue
                page_rows = [list(r) for r in page_tables[0]]
                # Recover the State column from text position before this
                # page's rows are handed off to process_and_save().
                page_rows = inject_state_into_rows(page, page_rows, state_col_idx=0)
                all_rows.extend(page_rows)
 
            process_and_save(all_rows, filename)
```
